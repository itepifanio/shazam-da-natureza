# Baixando o conjunto de dados

Acesse `https://datadryad.org/dataset/doi:10.5061/dryad.d2547d81z` e baixe manualmente o dataset completo (não há método fácil de baixar via API).

Após o donwload extraia os arquivos e mova para a pasta `data` disponível nesse repositório. As células a seguir validarão se o conteúdo do arquivo está seguindo o que é esperado pelos próximos notebooks.
