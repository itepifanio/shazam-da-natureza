# Particionando dados

Separando os dados em treinamento, validação e teste. 

Esse notebook irá usar um dos lugares de gravação, chamado de `Recording_4`, como conjunto de teste e os demais serão usados para treinamento/validação.

```python
from pathlib import Path

import sklearn
import pandas as pd
```

```python
clip_labels_file = Path("clip_labels.csv")
assert clip_labels_file.exists(), "O clip_labels.csv não foi encontrado. Execute o notebook 01_visualizando.ipynb para criar esse arquivo"
```

```python
clip_labels = pd.read_csv(clip_labels_file, index_col=["file", "start_time", "end_time"])
clip_labels.head()
```

```python
mask = clip_labels.reset_index()["file"].apply(lambda x: "Recording_4" in x).values
test_set = clip_labels[mask]
test_set.head()
```

```python
test_set.to_csv("test_set.csv")
```

Agora que separamos os dados de teste, os demais arquivos serão utilizados para treino e validação.

```python
train_and_val_set = clip_labels.drop(test_set.index)
```

```python
train_df, valid_df = sklearn.model_selection.train_test_split(
    train_and_val_set, test_size=0.1, random_state=0
)
```

```python
train_df.to_csv("train.csv")
```

```python
valid_df.to_csv("valid.csv")
```
